## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Research question:** Which pages should be reviewed first for content refresh or improvement based on information that was observable before the decision?

**Decision:** which content items receive scarce human review capacity first.

**Human action:** a reviewer may refresh, expand, protect, prune, or monitor a page after checking the evidence.

**Cost of error:** a false positive wastes editorial effort; a false negative can leave a useful opportunity or risk unreviewed. The project therefore optimizes prioritization, not guaranteed traffic recovery.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```
import pandas as pd
import json

# The supplied starter slice is intentionally used only for Week 1–2 discovery.
url = (
    "https://raw.githubusercontent.com/"
    "sahil2003-ai/flyrank-ml-intern-assignments/"
    "main/data/raw/content_refresh_anonymized.csv"
)
df = pd.read_csv(url)

summary = {
    "rows": int(len(df)),
    "columns": int(len(df.columns)),
    "median_impressions_90d": float(df["impressions_90d"].median()),
    "median_sessions_90d": float(df["sessions_90d"].median()),
    "trend_counts": df["trend_direction"].value_counts(dropna=False).to_dict(),
}

print(f"Dataset shape: {df.shape}")
print(f"Median 90-day impressions: {summary['median_impressions_90d']:,.1f}")
print(f"Median 90-day sessions: {summary['median_sessions_90d']:,.1f}")
print("\nTrend distribution:")
print(df["trend_direction"].value_counts(dropna=False))

out = Path(PROJECT_DIR) / "work" / "outputs" / "w01_summary.json"
out.write_text(json.dumps(summary, indent=2, default=str), encoding="utf-8")
print("\nSaved:", out)
```
