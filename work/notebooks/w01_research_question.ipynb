## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## Research question

**Which pages should be reviewed first for content refresh or improvement based on observable search visibility, engagement, freshness, and content signals?**

### Decision

The decision is which pages a content team should prioritize for human review when review capacity is limited.

### Who acts on the decision?

A content or SEO reviewer would use the ranked output to decide which pages deserve attention first.

### Action

Depending on the evidence, the reviewer could choose to **refresh, expand, protect, prune, or monitor** a page.

### Cost of a wrong call

A false positive could cause the team to spend time and resources changing a page that did not need intervention. A false negative could cause an important page with a genuine opportunity or risk to be overlooked. Therefore, the goal is to improve review prioritization rather than guarantee that a page will recover after an intervention.
